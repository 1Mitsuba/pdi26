## Negativo de una imagen 

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread("imagen.jpg",cv2.IMREAD_GRAYSCALE)
img_neg= 255 - img

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title(f'Original ')
plt.subplot(1, 2, 2)
plt.imshow(img_neg, cmap="gray")
plt.title('negativo')
plt.tight_layout()
plt.show()
```

## Negativo a color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread("imagen.jpg")
img_neg= 255 - img

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title(f'Original ')
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_neg,cv2.COLOR_BGR2RGB))
plt.title('negativo')
plt.tight_layout()
plt.show()
```

## Contraste

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def contraste(img,factor=1.0):
    img=img.astype(np.float32)
    media=np.mean(img)
    img_contraste=media+(img-media)*factor
    img_contraste=np.clip(img_contraste,0,255).astype(np.uint8)
    return img_contraste
img=cv2.imread("imagen.jpg",0)
contrast=9
img_fin=contraste(img,contrast)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title(f'Original ')
plt.subplot(1, 2, 2)
plt.imshow(img_fin, cmap="gray")
plt.title('contraste')
plt.tight_layout()
plt.show()
```

## Contraste a color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def contraste(img,factor=1.0):
    img=img.astype(np.float32)
    for i in range(3):
        media=np.mean(img[:,:,i])
        img[:,:,i]=media+(img[:,:,i]-media)*factor
        img[:,:,i]=np.clip(img[:,:,i],0,255)
    return img.astype(np.uint8)
img=cv2.imread("imagen.jpg")
contrast=3
img_fin=contraste(img,contrast)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title(f'Original ')
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_fin,cv2.COLOR_BGR2RGB))
plt.title('contraste')
plt.tight_layout()
plt.show()
```

## Brillo

## Brillo a color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def brillo (img,brillo=0):
    img_brillo=img.astype(np.float32)+brillo
    img_brillo=np.clip(img_brillo,0,255).astype(np.uint8)
    return img_brillo

img=cv2.imread('imagen.jpg')
bril=-70
img_fin=brillo(img,bril)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('imagen original')
plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB))
plt.title('imagen con brillo')
plt.show()
```
