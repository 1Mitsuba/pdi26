# Actividad de Laboratorio - Procesando Imágenes

**Objetivo:** Aplicar cv2 para cargar, procesar y guardar imágenes.

- Cargar 3 imágenes en formato BGR usando `cv2.imread`
- Mostrar cada imagen convertida a RGB usando Matplotlib
- Convertir a escala de grises con `cv2.cvtColor`
- Guardar imagen BGR y escala de grises
- Crear función para mostrar las imágenes
- Mostrar en dos modos de subplot: canales en escala de grises, color y color falso
- Cada imagen en segmento de código separado

```python
# Importar librerías necesarias
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams["figure.figsize"] = [10, 5]
```

## Función para mostrar imágenes en dos modos de subplot

```python
def mostrar_imagen_variantes(img_bgr, titulo=""):
    """
    Muestra la imagen en dos modos de subplot diferentes:
    - Modo 1: canales RGB en escala de grises
    - Modo 2: imagen original color, escala de grises y color falso
    """
    if img_bgr is None:
        print(f"Imagen {titulo} no encontrada")
        return
    
    # Convertir BGR a RGB para matplotlib
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    
    # Convertir a escala de grises
    img_gris = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    
    # Color falso (aplicar colormap)
    img_falso_color = cv2.applyColorMap(img_gris, cv2.COLORMAP_JET)
    img_falso_color_rgb = cv2.cvtColor(img_falso_color, cv2.COLOR_BGR2RGB)
    
    # MODO 1: Mostrar canales RGB en escala de grises
    fig, axs = plt.subplots(1, 3, figsize=(15, 5))
    
    fig_img = axs[0].imshow(img_rgb[..., 0], cmap="gray")
    fig.colorbar(fig_img, ax=axs[0], shrink=0.5)
    axs[0].set_title("Canal Rojo")
    axs[0].axis('off')
    
    fig_img = axs[1].imshow(img_rgb[..., 1], cmap="gray")
    fig.colorbar(fig_img, ax=axs[1], shrink=0.5)
    axs[1].set_title("Canal Verde")
    axs[1].axis('off')
    
    fig_img = axs[2].imshow(img_rgb[..., 2], cmap="gray")
    fig.colorbar(fig_img, ax=axs[2], shrink=0.5)
    axs[2].set_title("Canal Azul")
    axs[2].axis('off')
    
    plt.suptitle(f"{titulo} - Modo 1: Canales en escala de grises")
    plt.tight_layout()
    plt.show()
    
    # MODO 2: Mostrar imagen original, escala de grises y color falso
    fig, axs = plt.subplots(1, 3, figsize=(15, 5))
    
    axs[0].imshow(img_rgb)
    axs[0].set_title("Original (Color)")
    axs[0].axis('off')
    
    axs[1].imshow(img_gris, cmap="gray")
    axs[1].set_title("Escala de Grises")
    axs[1].axis('off')
    
    axs[2].imshow(img_falso_color_rgb)
    axs[2].set_title("Color Falso")
    axs[2].axis('off')
    
    plt.suptitle(f"{titulo} - Modo 2: Color, Grises y Color Falso")
    plt.tight_layout()
    plt.show()
```

## Imagen 1: Carito

```python
# Cargar imagen en formato BGR
img_carito = cv2.imread("Carito.jpg.jpeg")

if img_carito is None:
    print("ERROR: Carito.jpg.jpeg no encontrada")
else:
    # Guardar imagen en BGR en el ordenador
    cv2.imwrite("Carito_guardada_bgr.jpg", img_carito)
    
    # Convertir a escala de grises
    img_carito_gris = cv2.cvtColor(img_carito, cv2.COLOR_BGR2GRAY)
    
    # Guardar imagen en escala de grises
    cv2.imwrite("Carito_gris.jpg", img_carito_gris)
    
    # Mostrar imagen usando la función
    mostrar_imagen_variantes(img_carito, "Carito")
    
    print("Carito procesada y guardada exitosamente")
```
