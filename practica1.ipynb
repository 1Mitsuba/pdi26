# Práctica 1 - Ejercicios de lógica

Este notebook contiene las 7 aplicaciones solicitadas con un menú interactivo para ejecutar cada ejercicio.

Seleccione una opción en el menú (0 para salir).

```python
# --- Definiciones en español de funciones y utilidades ---
from collections import Counter
import re
import os
try:
    from IPython.display import clear_output
    _have_clear_output = True
except Exception:
    _have_clear_output = False

def limpiar_pantalla():
    if _have_clear_output:
        clear_output(wait=True)
    else:
        os.system('cls' if os.name == 'nt' else 'clear')

def pausa():
    try:
        input('\nPresione Enter para volver al menú...')
    except Exception:
        # en entornos donde input falle, ignorar
        pass

def factorial(numero: int) -> int:
    if numero < 0:
        raise ValueError('No se puede calcular factorial de un número negativo')
    resultado = 1
    for i in range(2, numero+1):
        resultado *= i
    return resultado

def es_primo(numero: int) -> bool:
    if numero <= 1:
        return False
    if numero <= 3:
        return True
    if numero % 2 == 0:
        return False
    i = 3
    while i * i <= numero:
        if numero % i == 0:
            return False
        i += 2
    return True

def max_y_min_n_numeros() -> tuple:
    while True:
        try:
            limite = int(input('Ingrese la cantidad de números a introducir (mínimo 3): '))
            if limite < 3:
                print('Debe ingresar al menos 3 números.')
                continue
            break
        except ValueError:
            print('Entrada inválida, intente de nuevo.')
    numeros = []
    for i in range(limite):
        while True:
            try:
                val = float(input(f'Número {i+1}: '))
                numeros.append(val)
                break
            except ValueError:
                print('Entrada inválida, ingrese un número.')
    return max(numeros), min(numeros)

def es_palindromo(cadena: str) -> bool:
    limpia = ''.join(re.findall(r"[A-Za-z0-9]", cadena)).lower()
    return limpia == limpia[::-1]

def introducir_7_y_contar_repeticiones() -> dict:
    numeros = []
    for i in range(7):
        while True:
            try:
                val = input(f'Ingrese número {i+1}: ')
                # permitir enteros y floats
                if '.' in val:
                    num = float(val)
                else:
                    num = int(val)
                numeros.append(num)
                break
            except ValueError:
                print('Entrada inválida, intente de nuevo.')
    conteos = Counter(numeros)
    return dict(conteos)

def es_anagrama(palabra1: str, palabra2: str) -> bool:
    limpia1 = ''.join(re.findall(r"[A-Za-z0-9]", palabra1)).lower()
    limpia2 = ''.join(re.findall(r"[A-Za-z0-9]", palabra2)).lower()
    return sorted(limpia1) == sorted(limpia2)

def ordenamiento_burbuja(lista):
    arr = lista[:]
    n = len(arr)
    for i in range(n):
        cambiado = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                cambiado = True
        if not cambiado:
            break
    return arr

# --- Menú interactivo en español ---
def mostrar_menu():
    while True:
        limpiar_pantalla()
        texto_menu = '\n'.join([
            '----- MENÚ DE EJERCICIOS -----',
            '1 - Calcular el factorial de un número entero',
            '2 - Determinar si un número es primo',
            '3 - Ingresar N números (≥3) y mostrar mayor y menor',
            '4 - Verificar si una cadena es palíndromo',
            '5 - Introducir 7 números y contar repeticiones',
            '6 - Verificar si dos palabras son anagramas',
            '7 - Ordenar números mediante el método burbuja',
            '0 - Salir'
        ])

        print('\n' + texto_menu)
        opcion = input('Seleccione una opción: ').strip()
        if opcion == '0':
            limpiar_pantalla()
            print('Saliendo.')
            break
        elif opcion == '1':
            try:
                n = int(input('Ingrese un entero no negativo: '))
                print(f'Factorial de {n} = {factorial(n)}')
            except ValueError as e:
                print('Entrada inválida:', e)
            pausa()
        elif opcion == '2':
            try:
                n = int(input('Ingrese un número entero: '))
                print(f'{n} es primo? {es_primo(n)}')
            except ValueError:
                print('Entrada inválida.')
            pausa()
        elif opcion == '3':
            mayor, menor = max_y_min_n_numeros()
            print(f'Mayor: {mayor} ; Menor: {menor}')
            pausa()
        elif opcion == '4':
            s = input('Ingrese la cadena: ')
            print(f'Es palíndromo? {es_palindromo(s)}')
            pausa()
        elif opcion == '5':
            conteos = introducir_7_y_contar_repeticiones()
            print('Conteos de los 7 números:')
            for num, cnt in conteos.items():
                print(f'  {num} -> {cnt} veces')
            pausa()
        elif opcion == '6':
            a = input('Primera palabra: ')
            b = input('Segunda palabra: ')
            print(f'Son anagramas? {es_anagrama(a,b)}')
            pausa()
        elif opcion == '7':
            entrada = input('Ingrese números separados por espacios: ').strip()
            try:
                lista = [float(x) if '.' in x else int(x) for x in entrada.split()]
                ordenada = ordenamiento_burbuja(lista)
                print('Lista ordenada:', ordenada)
            except Exception as e:
                print('Error al leer los números:', e)
            pausa()
        else:
            print('Opción inválida, intente de nuevo.')
            pausa()

# Para ejecutar el menú, el usuario debe ejecutar la siguiente llamada:
# mostrar_menu()

print('Notebook listo. Ejecute `mostrar_menu()` para empezar.')
```

```python
# Ejecutar el menú interactivo (ejecuta esta celda para comenzar)
mostrar_menu()
```
