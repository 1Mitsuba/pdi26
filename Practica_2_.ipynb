## Crear una imagen RGB con 4 cuadrantes en rojo, verde, azul y negro

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams["figure.figsize"]=[10,5]

size_imagen = 100

size_cuadrante = size_imagen // 2  

# Cuadrante 1: ROJO 
R_rojo = np.ones((size_cuadrante, size_cuadrante)) * 255
G_rojo = np.ones((size_cuadrante, size_cuadrante)) * 0
B_rojo = np.ones((size_cuadrante, size_cuadrante)) * 0
cuadrante_rojo = np.dstack((R_rojo, G_rojo, B_rojo))

# Cuadrante 2: VERDE 
R_verde = np.ones((size_cuadrante, size_cuadrante)) * 0
G_verde = np.ones((size_cuadrante, size_cuadrante)) * 255
B_verde = np.ones((size_cuadrante, size_cuadrante)) * 0
cuadrante_verde = np.dstack((R_verde, G_verde, B_verde))

# Cuadrante 3: AZUL 
R_azul = np.ones((size_cuadrante, size_cuadrante)) * 0
G_azul = np.ones((size_cuadrante, size_cuadrante)) * 0
B_azul = np.ones((size_cuadrante, size_cuadrante)) * 255
cuadrante_azul = np.dstack((R_azul, G_azul, B_azul))

# Cuadrante 4: NEGRO 
R_negro = np.ones((size_cuadrante, size_cuadrante)) * 0
G_negro = np.ones((size_cuadrante, size_cuadrante)) * 0
B_negro = np.ones((size_cuadrante, size_cuadrante)) * 0
cuadrante_negro = np.dstack((R_negro, G_negro, B_negro))

fila_superior = np.hstack((cuadrante_rojo, cuadrante_verde))

fila_inferior = np.hstack((cuadrante_azul, cuadrante_negro))

imagen = np.vstack((fila_superior, fila_inferior))

imagen = np.array(imagen, dtype=int)

# Mostrar la imagen
plt.figure()
plt.imshow(imagen)
plt.title("Imagen RGB con 4 cuadrantes: Rojo, Verde, Azul y Negro")
plt.show()
```
