# COMBINAR FRONTERAS, REGIONES Y CONTEO

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import measure

def contar_elementos(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    _, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

    e_img = measure.label(binary, connectivity=2)
    num_obj = np.max(e_img)
    return num_obj, e_img, binary

img = cv2.imread('pikachu.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# combino tecnicas
blur = cv2.GaussianBlur(gray, (7,7), 0)
_, binary = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# contar
binary_bgr = cv2.cvtColor(binary, cv2.COLOR_GRAY2BGR)
conteo, e_img, binary = contar_elementos(binary_bgr)

# identificar por colores
color = np.random.randint(0, 255, size=(conteo + 1, 3), dtype=np.uint8)
color[0] = [0, 0, 0]
colorear = color[e_img]

print("Numero de regiones:", conteo)

for rid in range(1, conteo + 1):
    mask = (e_img == rid).astype(np.uint8) * 255
    region_color = np.zeros_like(colorear)
    region_color[mask == 255] = colorear[mask == 255]

    pix = np.sum(e_img == rid)
    c = color[rid]
    print(f"Region {rid}: {pix} Color de los pixecel RGB = {c.tolist()}")
```
